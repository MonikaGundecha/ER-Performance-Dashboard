```python
base_path = "/Volumes/workspace/default/hospital_raw/raw/raw/"
```

```python
# Read raw files
patient_df = spark.read.csv(base_path + "raw_patient_master.csv", header=True, inferSchema=True)
provider_df = spark.read.csv(base_path + "raw_provider_roster.csv", header=True, inferSchema=True)
dept_df = spark.read.csv(base_path + "raw_department_reference.csv", header=True, inferSchema=True)
visit_df = spark.read.csv(base_path + "raw_er_event_log.csv", header=True, inferSchema=True)
followup_df = spark.read.csv(base_path + "raw_followup_log.csv", header=True, inferSchema=True)

# Add ingestion timestamp
from pyspark.sql.functions import current_timestamp

patient_df = patient_df.withColumn("ingestion_ts", current_timestamp())
provider_df = provider_df.withColumn("ingestion_ts", current_timestamp())
dept_df = dept_df.withColumn("ingestion_ts", current_timestamp())
visit_df = visit_df.withColumn("ingestion_ts", current_timestamp())
followup_df = followup_df.withColumn("ingestion_ts", current_timestamp())

# Save as Bronze tables
patient_df.write.mode("overwrite").format("delta").saveAsTable("bronze_patient_master")
provider_df.write.mode("overwrite").format("delta").saveAsTable("bronze_provider_roster")
dept_df.write.mode("overwrite").format("delta").saveAsTable("bronze_department_reference")
visit_df.write.mode("overwrite").format("delta").saveAsTable("bronze_er_event_log")
followup_df.write.mode("overwrite").format("delta").saveAsTable("bronze_followup_log")
```
